```python
%sql
CREATE CATALOG IF NOT EXISTS IDENTIFIER(:catalog);
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog + 'landing') COMMENT "landing zone for all the files ingestions";
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog + 'bronze')  COMMENT "schema for all the raw files stores";
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog + 'silver') COMMENT "schema for all the clean data";
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog + 'gold') COMMENT "schema for all the aggregated data";
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog + 'ops') COMMENT "schema for all the operations";
```

```python
%sql
CREATE VOLUME IF NOT EXISTS IDENTIFIER(:catalog + 'landing.raw');
CREATE VOLUME IF NOT EXISTS IDENTIFIER(:catalog + 'ops.pipeline_state');
```
